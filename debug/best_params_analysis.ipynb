```python
import pandas as pd

# Load the parquet file
df = pd.read_parquet('benchopt_run_2025-07-03_19h12m20_allparams.parquet')

# Filter for converged runs (stop_val == 0)
df_converged = df[df['stop_val'] == 0]

# For each dataset, select the converged run with the lowest time
best_time = (
    df_converged
    .sort_values('time')
    .groupby('data_name', as_index=False)
    .first()
)
best_time.to_csv('top_by_time.csv', index=False)

# For each dataset, select the converged run with the lowest objective value
best_obj = (
    df_converged
    .sort_values('objective_value')
    .groupby('data_name', as_index=False)
    .first()
)
```
